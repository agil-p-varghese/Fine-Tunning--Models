```python
import fitz #PyMuPdf
from docling.document_converter import DocumentConverter
```

```python
#target:"attention is all you need"
pdf_url="https://arxiv.org/pdf/1706.03762.pdf"
local_pdf="attention.pdf"
```

```python
#download the pdf locally
import urllib.request
urllib.request.urlretrieve(pdf_url,local_pdf)
print("Downloaded PDF successfully!!"+"="*20)
#test a industry grade parsing--docling
print("\n[Running Docling-layout parser . . .]")
converter=DocumentConverter()
docling_result=converter.convert(local_pdf)
docling_md=docling_result.document.export_to_markdown()

print("Docling output----------first(1000 chars)---------\n")
print(docling_md[10000:11000])
```

```python
# ==========================================
# ROBUST INSPECTION: DOCLING VS NAIVE (PAGE 6)
# ==========================================

print("--- 📄 DOCLING: FULL MARKDOWN EXPORT (Searching for Tables/Page content) ---")
# Let's print chunks of the markdown where tables usually appear
# (You can search for keywords like 'Table' or 'BLEU' in the docling_md)
for line in docling_md.split("\n"):
    if "Table" in line or "BLEU" in line or "perplexity" in line.lower():
        print(line)

print("\n" + "="*50)
print("--- 📄 NAIVE PYMUPDF: PAGE 6 TEXT (Raw Output) ---")
with fitz.open(local_pdf) as doc:
    page_6_naive = doc[5].get_text("text") # 0-indexed, page 6
    print(page_6_naive)
```

```python
#naive extraction(PYMUPDF)
print("\n"+"="*10+"\n [Runnign naive extractor......]"+"="*10)
naive_text=""
with fitz.open(local_pdf) as doc:
  #lets check page 3(formulas,tables)
  naive_text=doc[6].get_text("text")

print("------Naive PyMuPDF output ------------\n")
print(naive_text[:1000])
```
