```python
import torch
from peft import PeftModel
from transformers import AutoModelForCausalLM, AutoTokenizer

base_model_id = "Qwen/Qwen2.5-3B-Instruct"
adapter_path = "./support_bot_out"

print("Loading tokenizer...")
tokenizer = AutoTokenizer.from_pretrained(base_model_id)

print("Loading base model directly to GPU...")
base_model = AutoModelForCausalLM.from_pretrained(
    base_model_id,
    dtype=torch.bfloat16,
    device_map="cuda:0"
)

print("Loading your fine-tuned LoRA weights...")
model = PeftModel.from_pretrained(base_model, model_id=adapter_path)
model.eval()

# Test prompt formatted with the Alpaca instruction structure
prompt_text = (
    "Below is an instruction that describes a task, paired with an input that provides further context. "
    "Write a response that appropriately completes the request.\n\n"
    "### Instruction:\nRespond according to the support response policy.\n\n"
    "### Input:\ni have brought an item and its been 30 days since delivery,can i raise a request for return or refund??\n\n"
    "### Response:\n"
)

inputs = tokenizer(prompt_text, return_tensors="pt")
inputs = {k: v.to("cuda") for k, v in inputs.items()}

print("Generating response...")
with torch.no_grad():
    outputs = model.generate(
        **inputs,
        max_new_tokens=100,
        temperature=0.3,
        do_sample=True,
        pad_token_id=tokenizer.pad_token_id
    )

response = tokenizer.decode(outputs[0], skip_special_tokens=True)
print("\n--- Model Output ---")
print(response)
```
